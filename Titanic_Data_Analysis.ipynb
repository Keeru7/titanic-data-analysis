```python
df = df.drop(columns=["Cabin"])
```

```python
df["Age"] = df["Age"].fillna(df["Age"].median())
```

```python
df.isnull().sum()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sns.barplot(data=df, x="Sex", y="Survived")

plt.title("Survival Rate by Gender")
plt.xlabel("Gender")
plt.ylabel("Survival Rate")

plt.show()
```

```python
sns.barplot(data=df, x="Pclass", y="Survived")

plt.title("Survival Rate by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Survival Rate")

plt.show()
```

```python
sns.boxplot(data=df, x="Survived", y="Age")

plt.title("Age Distribution by Survival")
plt.xlabel("Survived (0 = No, 1 = Yes)")
plt.ylabel("Age")

plt.show()
```

```python
sns.scatterplot(data=df, x="Age", y="Fare", hue="Survived")

plt.title("Age vs Fare")
plt.xlabel("Age")
plt.ylabel("Fare")

plt.show()
```

```python
sns.heatmap(df.corr(numeric_only=True), annot=True)

plt.title("Correlation Heatmap")

plt.show()
```

```python
sns.boxplot(data=df, y="Fare")

plt.title("Fare Outlier Analysis")
plt.ylabel("Fare")

plt.show()
```

## Key Takeaways

- Survival rates were different between male and female passengers.
- Passenger class had a noticeable relationship with survival rate.
- Most passengers were within a common adult age range, while a smaller number were very young or older.
- Fare values varied considerably, with some passengers paying unusually high fares.
- The correlation heatmap shows relationships between numerical variables such as Pclass, Age, Fare, and Survived.
- The analysis helped identify patterns, relationships, and potential outliers before applying machine learning.

### Takeaway

The boxplot shows the age distribution of passengers who survived and those who did not. It also helps identify unusual age values in the dataset.
